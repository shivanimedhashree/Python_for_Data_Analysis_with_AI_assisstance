```python
""" Task 1: Single-Column Groupby Aggregation
Question: Calculate the total sales generated by each region. """

#code
task1_superstore = df_superstore.groupby('Region')['Sales'].sum().reset_index()
print(task1_superstore)
```

```python
""" Task2: Multi-Column Groupby Aggregation
Question: Calculate total sales and Profit broken down by Category and Sub-Category. """

#code
task2_superstore = df_superstore.groupby(['Category', 'Sub-Category'])[['Sales', 'Profit']].sum().reset_index()

print(task2_superstore.head(5))
```

```python
""" Task 3 - Multi-metric Aggregation with agg()
Question: Compute
- average Sales
- Minimum Profit
- Maximum Profit
for each Product category. """

#code
task3_superstore = df_superstore.groupby('Category').agg(
    Avg_sales=('Sales', 'mean'),
    Min_Profit=('Profit', 'min'),
    Max_Profit=('Profit', 'max')
).reset_index()
print(task3_superstore)
```

```python
""" Task 4 - Sorting Aggregate Summaries
Question: Identify top 5 Sub-Categories generating the highest cumulative Profit. """

#code
task4_superstore = (df_superstore.groupby('Sub-Category')['Profit']
                    .sum()
                    .reset_index()
                    .sort_values(by='Profit', ascending=False)
                    .head(5))
print(task4_superstore)
```

```python
""" Task 5 - Named Aggregations
Question: Calculate total Order Count, average Discount rate, and total Profit
          across Customer Segments. """

#code
task5_superstore = df_superstore.groupby('Segment').agg(
    Total_Order=('Row ID', 'count'),
    Avg_Discount=('Discount', 'mean'),
    Total_Profit=('Profit', 'sum')
).reset_index()
print(task5_superstore)
```

```python
""" Task 6 - Conditional Group Filtering
Question: Filter the dataset to isolate rows originating only from Cities whose
          total Sales exceed $50,000. """

#code
task6_superstore = df_superstore.groupby('City').filter(lambda x: x['Sales'].sum() > 50000)
print("Qualified Cities:", task6_superstore['City'].unique())
print("Retained Rows Ratio:", f"{len(task6_superstore)} / {len(df_superstore)}")
```

```python
""" Task 7 - Derived Aggregate Calculations
Question: Calculate
          - total sales
          - total profit
          - Profit Margin percentage (PRofit/Sales*100) per Ship Mode. """

#code
task7_superstore = df_superstore.groupby('Ship Mode').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
).reset_index()
task7_superstore['Profit_Margin_%'] = (task7_superstore['Total_Profit'] / task7_superstore['Total_Sales']) * 100
print(task7_superstore.sort_values(by='Profit_Margin_%', ascending=False))
```

```python
""" Task 8 - Extremum Selection with nlargest()
Question: Extract the top 5 individual orders based on gross Sales value. """

#code
task8_superstore = df_superstore.nlargest(5, 'Sales')[['Order ID', 'Category', 'Sub-Category',
                                                       'Sales', 'Profit']]
print(task8_superstore)
```

```python
""" Task 9 - Quantile Aggregation
Question: Find the 75th percentile (Q3) of order Sales across each Product Category. """

#code
task9_superstore = df_superstore.groupby('Category')['Sales'].agg(
    Q3_Sales=lambda x: x.quantile(0.75)
).reset_index()
print(task9_superstore)
```

```python
""" Task 10 - Frequency Counting via Groupby
Question: Count total order line items across combinations of Region and Customer Segment. """

#code
task10_superstore = (df_superstore.groupby(['Region', 'Segment'])
                     .size()
                     .reset_index(name='Order_Count')
                     .sort_values(by='Order_Count', ascending=False)
                     )
print(task10_superstore.head(5))

```

**Task 1:** Resample Daily Sales into Weekly Aggregate Sales.



```python
import matplotlib.pyplot as plt

# Convert 'Order Date' to datetime and set as index
df_superstore["Order Date"] = pd.to_datetime(df_superstore["Order Date"])
df_superstore_time = df_superstore.set_index("Order Date")

# Resample 'Sales' weekly ('W') using sum aggregation
weekly_sales = df_superstore_time["Sales"].resample("W").sum()

# Plot the line chart
plt.figure(figsize=(12, 6))
weekly_sales.plot(kind="line")
plt.title("Weekly Sales Trend (Superstore)")
plt.xlabel("Order Date")
plt.ylabel("Total Sales")
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Task 2:** Calculate and Plot 7-Day Moving Average of Sales.

```python
import matplotlib.pyplot as plt

# Ensure 'Order Date' is datetime and set as index. This was done in the previous cell
# df['Order Date'] = pd.to_datetime(df['Order Date'])
# df_superstore_time = df.set_index('Order Date')

# Aggregate daily sales (sum, in case there are multiple orders per day)
daily_sales = df_superstore_time["Sales"].resample("D").sum()

# Calculate the 7-day rolling mean of daily sales
rolling_mean_7_day = daily_sales.rolling(window=7).mean()

# Plot original daily sales and the 7-day rolling mean
plt.figure(figsize=(14, 7))
plt.plot(daily_sales, label="Daily Sales", alpha=0.7)
plt.plot(rolling_mean_7_day, label="7-Day Rolling Mean", color="red")
plt.title("Daily Sales and 7-Day Rolling Mean (Superstore)")
plt.xlabel("Order Date")
plt.ylabel("Total Sales")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Task 3:** Calculate Period-over-Period Daily Growth using .shift()


```python
daily_sales_df = daily_sales.to_frame()
daily_sales_df["Sales_Prev_Day"] = daily_sales_df["Sales"].shift(1)
daily_sales_df["Daily Pct Change"] = (
    daily_sales_df["Sales"] - daily_sales_df["Sales_Prev_Day"]
) / daily_sales_df["Sales_Prev_Day"]
display(daily_sales_df.head())
```

**Task 2:** Matplotlib 3D Surface Plot of Synthetic Sales-Profit Surface.


```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# Create a synthetic dataset for X, Y, and Z values
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# Define a function for the Z-axis (e.g., a 2D sine wave or a quadratic surface)
Z = np.sin(np.sqrt(X**2 + Y**2)) # Example: A ripple effect
# Z = X**2 + Y**2 # Another example: A parabolic surface

# Create a 3D plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surface = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

# Set labels and title
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title('Smooth 3D Surface Plot')

# Add a color bar which maps values to colors
fig.colorbar(surface, shrink=0.5, aspect=10, label='Z Value')

plt.tight_layout()
plt.show()
```

**Task 9:** Dual-Axis Interactive Graph Objects.

```python
import matplotlib.pyplot as plt
import squarify
import numpy as np

# Group by 'Sub-Category' and sum 'Sales'
subcategory_sales = df_superstore.groupby('Sub-Category')['Sales'].sum().reset_index()

# Sort for better visualization (optional)
subcategory_sales = subcategory_sales.sort_values(by='Sales', ascending=False)

# Prepare labels for the treemap
labels = subcategory_sales.apply(lambda x: f"{x['Sub-Category']}\n({x['Sales']:.2f})", axis=1)

# Generate a list of colors from the colormap
num_subcategories = len(subcategory_sales)
colors = [plt.cm.Spectral(i / float(num_subcategories)) for i in range(num_subcategories)]

# Create the treemap
plt.figure(figsize=(12, 8))
squarify.plot(
    sizes=subcategory_sales['Sales'],
    label=labels,
    alpha=.8,
    color=colors # Pass the list of generated colors
)

plt.title('Total Sales by Sub-Category', fontsize=16)
plt.axis('off') # Hide axes
plt.show()
```

**Task 10:** Multi-Component KPI Card and Category Pie Chart Layout

```python
import dash
from dash import dcc, html
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd

# Initialize the Dash app
app_kpi = dash.Dash(__name__)

# Calculate overall KPIs
total_sales = df_superstore['Sales'].sum()
total_profit = df_superstore['Profit'].sum()
profit_margin = (total_profit / total_sales) * 100 if total_sales != 0 else 0

# Data for Category Sales Pie Chart
category_sales = df_superstore.groupby('Category')['Sales'].sum().reset_index()

# Define the layout of the app
app_kpi.layout = html.Div([
    html.H1("Superstore Sales Overview"),

    # KPI Cards
    html.Div([
        html.Div([
            html.H3("Total Sales"),
            html.P(f"${total_sales:,.2f}", style={'fontSize': 24, 'fontWeight': 'bold'})
        ], style={'flex': 1, 'padding': 10, 'border': '1px solid #ddd', 'margin': 5, 'textAlign': 'center'}),

        html.Div([
            html.H3("Total Profit"),
            html.P(f"${total_profit:,.2f}", style={'fontSize': 24, 'fontWeight': 'bold'})
        ], style={'flex': 1, 'padding': 10, 'border': '1px solid #ddd', 'margin': 5, 'textAlign': 'center'}),

        html.Div([
            html.H3("Profit Margin"),
            html.P(f"{profit_margin:,.2f}%", style={'fontSize': 24, 'fontWeight': 'bold'})
        ], style={'flex': 1, 'padding': 10, 'border': '1px solid #ddd', 'margin': 5, 'textAlign': 'center'})
    ], style={'display': 'flex', 'justifyContent': 'space-around', 'marginBottom': 20}),

    # Category Sales Pie Chart
    html.Div([
        dcc.Graph(
            id='category-sales-pie-chart',
            figure=px.pie(
                category_sales,
                values='Sales',
                names='Category',
                title='Sales Distribution by Category',
                height=500
            )
        )
    ], style={'width': '100%', 'marginTop': 20})
])

# Run the app
if __name__ == '__main__':
    app_kpi.run(debug=True, mode='inline', port=8052) # Using a different port (8052) to avoid conflict
```
